## Section 1: Introduction to Jupyter Notebooks

### What is a Jupyter Notebook?

A Jupyter Notebook is an interactive document with the file extension `.ipynb`. Unlike a regular Python script (`.py`), a notebook lets you mix **code**, **narrative text**, and **output** (text, tables, charts) all in one place, organized into individual **cells** that run independently.

Think of it like a lab notebook: you write down what you're doing, run an experiment (a code cell), see the result immediately, and add notes around it — all without leaving the document.
- Web-based application for authoring documents that combine live-code with narrative text, equations and visualizations.
- Jupyter Notebooks are structured data that represent your code, metadata, content, and outputs. When saved to disk, the notebook uses the extension **.ipynb**, and uses a **JSON** structure. 
- Jupyter Notebook and its flexible interface extends the notebook beyond code to visualization, multimedia, collaboration, and more. 
- To running your code, it stores code and output, together with markdown notes, in an editable document called a notebook. When you save it, this is sent from your browser to the Jupyter server, which saves it on disk as a JSON file with a **.ipynb** extension.



- The Jupyter server is a communication hub. The browser, notebook file on disk, and kernel cannot talk to each other directly. They communicate through the Jupyter server. 
- The Jupyter server, not the kernel, is responsible for saving and loading notebooks, so you can edit notebooks even if you don’t have the kernel for that language—you just won’t be able to run code. 
- The kernel doesn’t know anything about the notebook document: it just gets sent cells of code to execute when the user runs them.
- A **kernel** is a “computational engine” that executes the code contained in a notebook document.
- A **cell** is a container for text to be displayed in the notebook or code to be executed by the notebook’s kernel.
### Why Jupyter is Used in AI/ML Work

- **Iterative experimentation** — test an idea in one cell without re-running your entire program
- **Immediate feedback** — see outputs, errors, and results right below the code that produced them
- **Inline visualization** — charts, tables, and images appear directly in the document
- **Documentation alongside code** — explain *why* you did something next to the code that does it
- **Industry standard** — used across research, prototyping, and even some production pipelines

### Anatomy of a Notebook

A notebook is made up of:

- **Cells** — the building blocks; each cell is either a **Code cell** or a **Markdown cell**
- **Kernel** — the "engine" running in the background that executes your code and remembers variables
- **Toolbar** — buttons to run cells, add/delete cells, save, and restart the kernel
- **Execution count** — the number in brackets (e.g., `[3]`) next to a code cell showing the order it was run in

### Code Cells vs Markdown Cells

- **Code cells** run Python (or another language) and show output directly below
- **Markdown cells** render formatted text — headings, bold, italics, lists, even math

You can switch a cell's type using keyboard shortcuts: press `Esc` then `M` for Markdown, or `Esc` then `Y` for Code.

- It supports **bold text**, *italic text*, and bullet lists
- It can also render simple math like $E = mc^2$
- Markdown cells never produce "output" — they simply render as formatted text

````python
# This is a Code cell example
message = "Hello from a code cell!"
print(message)
````

### Running Cells & Execution Order

Common ways to run a cell:

- **Shift + Enter** — run the current cell and move to the next one
- **Ctrl + Enter** — run the current cell and stay on it
- **Alt + Enter** — run the current cell and insert a new one below

Each time you run a code cell, its execution count (the number in `[ ]`) updates. This number tells you the **order cells were actually run in** — which is not always top-to-bottom!

**Why this matters:** running cells out of order can cause confusing bugs. Try the example below to see this in action.

````python
# Run this cell first
x = 10
print("x is:", x)
````

````python
# Now run this cell
x = x + 5
print("x is now:", x)
````

If you run the cell above **multiple times** without re-running the first one, `x` keeps growing by 5 each time — even though the code looks like it should always produce the same answer. This is exactly the kind of confusing bug that execution order can cause. Try running the second cell 2-3 times in a row and watch the value change.

### The Kernel — What It Is & Why It Matters

The **kernel** is the background process that actually runs your code and holds all your variables in memory. As long as the kernel is running, variables you create in one cell are available in every other cell — in any order.

This is powerful, but it also means your notebook's *visible* code and its *actual* memory state can drift apart if you're not careful. **Restarting the kernel** clears all memory and gives you a clean slate — this is the single most useful "fix it and try again" button in Jupyter.

Let's demonstrate kernel memory persisting across cells:

````python
# Create a variable
favourite_number = 42
print("Stored favourite_number:", favourite_number)
````

Now, without touching the cell above, run the cell below. Notice `favourite_number` is still available — the kernel remembered it.

````python
# This cell uses a variable created in a previous cell
print("Recalling favourite_number:", favourite_number)
````

**Try this:** Go to the menu and select **Kernel → Restart Kernel** (or the restart button in the toolbar). Then run *only* the cell above (without re-running the one that created `favourite_number`). You should get a `NameError` — proof that restarting the kernel wipes all stored variables. This is expected behavior, not a bug!

## Section 3: Virtual Environment Walkthrough  *(Hands-on Lab I)*

A virtual environment is an isolated Python installation dedicated to one project, so its packages don't conflict with any other project's packages.

**Important:** true virtual environment creation happens in your **terminal**, not inside a notebook cell (a notebook already runs inside a single Python environment/kernel). Below is the walkthrough you'll follow in your terminal, plus a notebook-safe demonstration of package installation.

### Terminal walkthrough (run these in your terminal, not this notebook)

```bash
# 1. Create a virtual environment named "env"
python -m venv env

# 2. Activate it
# On Windows:
env\Scripts\activate
# On macOS/Linux:
source env/bin/activate

# 3. Confirm you're inside the virtual environment
which python   # (macOS/Linux) or `where python` (Windows)

# 4. Install a package inside the isolated environment
pip install numpy

# 5. Deactivate when done
deactivate
```

### Notebook-safe demonstration

Inside a notebook (or Colab), we don't create a new venv, but we can still see how `!pip install` isolates packages **within the current kernel's environment**:

````python
# Install a small, safe package to observe the install process
!pip install --quiet rich
print("Installed 'rich' into the current environment.")
````

````python
# Confirm the package is now available
from importlib.metadata import version

print("Rich version:", version("rich"))
````

````python
!pip show rich
````

````python
!uv pip show rich
````

**Key takeaway:** in your terminal, `python -m venv env` + activation creates true isolation between projects. Inside a notebook/Colab session, you're already working inside one shared environment — which is exactly why keeping track of *what's installed where* matters so much in AI engineering.

## Section 6: API Keys & Environment Variables

Many GenAI applications rely on hosted models (like OpenAI, Groq, or Anthropic) accessed through an **API key**. Never write your API key directly into your code — instead, use environment variables or a secure prompt.

**We will not use a real API key in this notebook** — this section only demonstrates the *safe pattern* for handling one.

````python
import os

# Safe pattern 1: reading a key from an environment variable
# (You would normally set this in your terminal or a .env file, never hard-code it)
api_key = os.environ.get("MY_API_KEY")

if api_key:
    print("API key loaded from environment variable.")
else:
    print("No API key found in environment variables (expected in this demo).")
````

````python
# Safe pattern 2: securely prompting for a key at runtime (input is hidden as you type)
from getpass import getpass

# Uncomment the line below to try it interactively:
api_key = getpass("Enter your API key: ")
print("getpass() hides your input so the key never appears in the notebook or its output.")
````

**Best practices:**
- Store real keys in environment variables or a `.env` file (added to `.gitignore`)
- Never print an API key or commit it to a shared/public repository
- Rotate keys immediately if one is ever accidentally exposed

## Section 8: Running Your First LLM Application

Let's put everything together and run a real transformer model. We'll use `distilgpt2`, a small, fast text-generation model that's ideal for a first hands-on run — no API key required, since it downloads and runs locally (or in your Colab session).

````python
from transformers import pipeline

# Load a small pretrained model for text generation
generator = pipeline("text-generation", model="distilgpt2")
print("Model loaded successfully.")
````

````python
# Run prompt-based generation
prompt = "The future of AI engineering is"

output = generator(
    prompt,
    max_length=40,
    num_return_sequences=1,
    truncation=True,
)

print(output[0]["generated_text"])
````

**What just happened?**
1. We loaded a tokenizer and model behind the scenes using `pipeline()`
2. Our prompt was converted into tokens the model understands
3. The model generated new tokens one at a time, which were converted back into readable text

Try changing the `prompt` text above and re-running the cell to see how the output changes. This experimentation loop — prompt, run, observe, adjust — is the core workflow of working with LLMs.

## Section 9: Troubleshooting Playground

Environment errors are a normal part of AI engineering — the key skill is recognizing common patterns quickly. Let's deliberately trigger two common issues and see how to fix them.

### Issue 1: Requesting an unreasonably large output (batch/size-style error)

Asking a small model to generate far more tokens than makes sense for a demo can slow things down dramatically or hit memory limits on constrained systems — the same underlying issue as an out-of-memory (OOM) error, just at a smaller scale here.

````python
# This deliberately requests a very large max_length to illustrate the issue
try:
    output = generator(
        "This is a test of resource limits:",
        max_length=100000,  # unreasonably large for a quick demo
        truncation=True,
    )
    print(output[0]["generated_text"][:200], "...")
except Exception as e:
    print("Error encountered:", type(e).__name__, "-", e)
````

**The fix:** reduce `max_length` to a reasonable number for your use case and available memory (e.g., 40-100 for a quick demo). In real training/inference on larger models, the equivalent fix is reducing **batch size** or using a smaller model when you hit a genuine OOM error.

### Issue 2: Package version mismatch

A common real-world error looks like this when two libraries expect incompatible versions of a shared dependency:

```
ERROR: pip's dependency resolver does not currently take into account all the packages that are installed...
package-a 2.1.0 requires shared-lib>=4.0, but you have shared-lib 3.2.1 which is incompatible.
```

**How to read this error:**
1. Identify the two packages involved (`package-a` and `shared-lib` here)
2. Check which version `package-a` needs (`>=4.0`)
3. Either upgrade `shared-lib` to a compatible version, or downgrade `package-a` to a version compatible with your current `shared-lib`

**The fix, in practice:**
```bash
pip install shared-lib --upgrade
# or, if that breaks something else:
pip install package-a==<older-compatible-version>
```

**Best practice:** when this happens mid-project, it's often faster to start a fresh virtual environment and reinstall only what you need, rather than patching version-by-version.

### Recap: Troubleshooting Checklist

When something breaks, work through this order:

1. **Read the full error message** — the actual cause is usually in the last few lines
2. **Check versions** — `pip show <package>` to see what's installed vs. what's required
3. **Restart the kernel** — clears any stale state before retrying
4. **Search the exact error text** — most environment errors are common and well-documented
5. **When in doubt, start fresh** — a new virtual environment often resolves conflicts faster than patching one-by-one

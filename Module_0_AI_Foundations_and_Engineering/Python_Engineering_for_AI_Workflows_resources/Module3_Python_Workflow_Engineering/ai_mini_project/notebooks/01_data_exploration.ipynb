## Class balance

Checking how many rows belong to each label — an important thing to know
before training any classifier.


```python
df["label"].value_counts()
```

## Feature correlation with the label

A simple way to see which features look most related to the target,
before building any model.


```python
correlations = df.corr(numeric_only=True)["label"].sort_values(ascending=False)
correlations
```
